```python
from google import genai
from google.genai import types
```

```python

client=genai.Client(
    vertexai=True,
    project="empyrean-aurora-469815-n4",
    location="us-west1",
    http_options=types.HttpOptions(api_version="v1")
)
```

```python
query="Explain the difference between machine learning and deep learning in 2 sentences."
model="gemini-2.5-flash"

response = client.models.generate_content(
    model=model,
    contents=query
)
print(response.text)
```
